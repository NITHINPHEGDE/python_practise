```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
x=np.array([2023,2024,2025,2026])
y1=np.array([12,13,23,26])
y2=np.array([24,23,21,25])
plt.plot(x,y1,marker=".",markerfacecolor="red",markeredgecolor="cyan",linestyle="dashed",linewidth=4,markersize=30)
plt.plot(x,y2,marker="*",markerfacecolor="blue",markeredgecolor="purple",linestyle="solid",linewidth=3,markersize=10)
plt.show()
```

```python
#or
styles=dict(marker=".",markerfacecolor="red",markeredgecolor="cyan",linestyle="dashed",linewidth=4,markersize=30)
plt.plot(x,y2,**styles)
```

```python
plt.title("Class Size",fontsize=25,fontfamily="Arial",fontweight="bold",color="pink")
plt.xlabel("Students",fontsize=25,fontfamily="Arial",fontweight="bold",color="red")
plt.ylabel("years",fontsize=25,fontfamily="Arial",fontweight="bold",color="green")
plt.xticks(x)
plt.tick_params(axis="both",color="pink")
plt.plot(x,y1)
plt.plot(x,y2)
plt.show()
```

```python
plt.grid(linewidth=2,color="grey",linestyle="dashed") #axix="x" (or) axis="y"
plt.plot(x,y1)
plt.show()
```

```python
categoris=["whey","chicken","eggs","milk"]
protien=np.array([23,32,30,12])
plt.title("per 100g")
plt.xlabel("protien")
plt.ylabel("source")
plt.bar(categoris,protien)#plt.barh if we want horizontal
```

```python
categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values= np.array([300, 250, 275, 225])
colors= ["red", "yellow", "blue", "green"]

plt.pie(values, labels=categories,
        autopct="%1.1f%%",
        shadow=True,
        colors=colors,
        explode=[0, 0,0,0.1])

plt.title("Bro Code College")

plt.show()
```

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]) # Hours studied
y1= np.array([55,60,65,62,68,70,75,78,82,85,87])# Grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) # Hours studied
y2 = np.array([50,58,65,70,72,78,83,88,92,81,92]) # Grades
plt.scatter(x1,y1, color="skyblue",
alpha = 0.5,
s = 200,
label="Class A")

plt.scatter(x2, y2, color="red",
alpha = 0.5,
s = 200,
label="Claşs B")
plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grade")

plt.legend()
plt.show()
```

```python
x=np.array([1,2,3,4,5])

figure, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, x*2, color="red")
axes[0, 0].set_title("x*2")

axes[0, 1].plot(x,x ** 2, color="blue")
axes[0, 1].set_title("x ** 2")

axes[1,0].plot(x, x ** 3, color="green")
axes[1,0].set_title("x ** 3")

axes[1,1].plot(x, x ** 4, color="purple") 
axes[1, 1].set_title("x ** 4")

plt.tight_layout()
plt.show()
```
