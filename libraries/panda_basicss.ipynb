```python
import pandas as pd
```

```python
#series is a single column of data
data=[101,102,102]
series=pd.Series(data)
print(series)
```

```python
#series with index

data=[101,102,102]
series=pd.Series(data,index=["a","b","c"])
print(series)
```

```python
data=[101,102,102]
series=pd.Series(data,index=["a","b","c"])
print(series.loc["b"])#accessing with index

```

```python
data=[101,102,102]
series=pd.Series(data,index=["a","b","c"])
series.loc["b"]=500#changing value
print(series.loc["b"])#accessing with index
```

```python
data=[101,102,102]
series=pd.Series(data,index=["a","b","c"])
print(series.iloc[2])#with respect to integer indexing
```

```python
data=[101,102,102,500,600]
series=pd.Series(data,index=["a","b","c","d","e"])
print(series[series>200])
```

```python
#another format
calorie ={"day1":1750,"day2":1772,"day3":1238}
series=pd.Series(calorie)
print(series)
```

```python
calorie ={"day1":1750,"day2":1772,"day3":1238}
series=pd.Series(calorie)
series.loc["day3"]+=500
print(series)
```

```python
#adding a new coloumn
data ={
  "name":["nithin","shubha","likith","navtej"],
  "age":[30,20,21,22]
}
df=pd.DataFrame(data,index=["emp1","emp2","emp3","emp4"])
df["job"]=["engineer","designer","real-estate","business"]
print(df)
```

```python
#adding a new row
data ={
  "name":["nithin","shubha","likith","navtej"],
  "age":[30,20,21,22]
}
df=pd.DataFrame(data,index=["emp1","emp2","emp3","emp4"])
df["job"]=["engineer","designer","real-estate","business"]
new_row=pd.DataFrame([{"name":"vikas","age":23,"job":"spiritual-guide"}],index=["emp5"])
df=pd.concat([df,new_row])
print(df)
```

```python
#exercise
df=pd.read_csv("pokemon.csv",index_col="Name")
poke = input("enter pokemon name")
try:
  print(df.loc[poke])
except KeyError:
  print(f"{poke} not found")
```

```python
#filtering means keeping only those rows that we need
tall_pokemon=df[df["Height"]>=2]
print(tall_pokemon)
```

```python
#print fire type and flying type
ff_poke=df[(df["Type1"]=="Fire")&
           (df["Type2"]=="Flying")]
print(ff_poke)
```

```python
#agregate functions
print(df.mean(numeric_only=True))#if we wanted for single col print(df["Height"].mean())
print(df.sum(numeric_only=True))
print(df.min(numeric_only=True))
print(df.max(numeric_only=True))
```

```python
#grouping
group=df.groupby("Type1")
print(group["Height"].sum())
```

```python
# 1st step drop irrelavant coloumns
df=df.drop(columns=["No","Weight"])
print(df)
```

```python
#handle missing data by replacing or removing
df.dropna(subset=["Type2"])
```

```python
#by replacing
df=df.fillna({"Type2":"hybrid"})
print(df)
```

```python
df["Type1"]=df["Type1"].replace({"Grass":"GRASS"})
print(df)
```
