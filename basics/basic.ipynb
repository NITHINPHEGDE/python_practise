```python
#basic string operations



string ="Hello world"
print(len(string))#length of a string

print(string.index("o"))#indexing

print(string.count("l"))#count of a character

print(string[3:7])#string indexing

print(string[3:7:2])#string [start:stop:step]

print(string[::-1])#reversing

print(string.upper())
print(string.lower())  #changing case

aftersplit = string.split(" ")
print(aftersplit)#splitting

```

```python
#conditions if else elif
# One program to understand conditions in Python

age = 20
marks = 72
has_id = True

# 1. Simple if
if age >= 18:
    print("You are an adult")

# 2. if–else
if marks >= 40:
    print("You passed")
else:
    print("You failed")

# 3. if–elif–else
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "D"

print("Grade:", grade)

# 4. Logical operators (and / or / not)
if age >= 18 and has_id:
    print("You are allowed to enter")

# 5. Nested condition
if age >= 18:
    if marks >= 60:
        print("Eligible for scholarship")
    else:
        print("Not eligible for scholarship")

# 6. Ternary condition (one-line if–else)
status = "Pass" if marks >= 40 else "Fail"
print("Status:", status)
```

```python
# Dictionary 
people = {
    "Sridhar": {
        "age": 25,
        "city": "Bangalore",
        "hobbies": ["gym", "coding", "gaming"]
    },
    "Anita": {
        "age": 23,
        "city": "Mumbai",
        "hobbies": ["reading", "traveling"]
    },
    "Ramesh": {
        "age": 30,
        "city": "Delhi",
        "hobbies": ["cooking", "cycling"]
    }
}

# Accessing details
print(people["Sridhar"]["city"])       # Output: Bangalore
print(people["Anita"]["hobbies"][0])   # Output: reading

# Looping through everyone
for person, details in people.items():
    print(f"{person} lives in {details['city']} and is {details['age']} years old.")
    print(f"Hobbies: {', '.join(details['hobbies'])}")
    print("---")
```
