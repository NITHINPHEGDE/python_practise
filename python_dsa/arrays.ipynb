```python
#python program for second largest element in an array

arr = list(map(int, input("Enter array elements: ").split()))

largest = float('-inf')
second_largest = float('-inf')

for i in arr:
    if i > largest:
        second_largest = largest
        largest = i

    elif i > second_largest and i != largest:
        second_largest = i

if second_largest == float('-inf'):
    print("Second largest element doesn't exist")
else:
    print("Second largest:", second_largest)
```

```python
#move all the zeroes to the end

arr = list(map(int, input("Enter array elements: ").split()))

s = 0
l = len(arr) - 1

while s < l:

    if arr[s] == 0:

        while s < l and arr[l] == 0:
            l -= 1

        arr[s], arr[l] = arr[l], arr[s]

    s += 1

print(arr)
```

```python
#find all the duplicates in an array
arr = list(map(int, input("Enter array elements: ").split()))

seen = set()
duplicates = set()

for i in arr:
    if i in seen:
        duplicates.add(i)
    else:
        seen.add(i)

print("Duplicate elements:", list(duplicates))
```

```python
#check if the two strins ar anagram
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

if len(str1) != len(str2):
    print("Not Anagram")

else:
    freq1 = {}
    freq2 = {}

    for i in str1:
        freq1[i] = freq1.get(i, 0) + 1

    for i in str2:
        freq2[i] = freq2.get(i, 0) + 1

    if freq1 == freq2:
        print("Anagram")
    else:
        print("Not Anagram")
```

```python
#frequency of characters

string = input("Enter a string: ")

frequency = {}

for i in string:
    frequency[i] = frequency.get(i, 0) + 1

print(frequency)

# Step 2: Find the most frequent character
max_count = 0
max_char = ""

for i in frequency:
    if frequency[i] > max_count:
        max_count = frequency[i]
        max_char = i

print("Most frequent character:", max_char)
print("Frequency:", max_count)

```

```python
#merge two sorted arrays

def merge(arr1, arr2):
    l1 = len(arr1)
    l2 = len(arr2)

    s1 = 0
    s2 = 0

    arr = []

    while s1 < l1 and s2 < l2:
        if arr1[s1] <= arr2[s2]:
            arr.append(arr1[s1])
            s1 += 1
        else:
            arr.append(arr2[s2])
            s2 += 1

    while s1 < l1:
        arr.append(arr1[s1])
        s1 += 1

    while s2 < l2:
        arr.append(arr2[s2])
        s2 += 1

    return arr


print(merge([1, 3, 5], [2, 4, 6]))
```
