```python
import numpy as np
```

```python
X=np.array([
    [800, 2, 1],
    [1000, 2, 2],
    [1200, 3, 2],
    [1500, 3, 2],
    [1800, 4, 3],
    [2000, 4, 3]
],dtype=float)
y = np.array([
    [120000],
    [150000],
    [180000],
    [220000],
    [260000],
    [300000]
], dtype=float)
```

```python
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X = (X - X_mean) / X_std
y_mean = y.mean()
y_std = y.std()
y = (y - y_mean) / y_std
```

```python
W1 = np.random.randn(3, 16) * 0.01#three inputs 16 neuron
b1 = np.zeros((1, 16))#one bias per neuron
```

```python
W2 = np.random.randn(16, 1) * 0.01
b2 = np.zeros((1, 1))
```

```python
def relu(Z):
    return np.maximum(0, Z)

def relu_derivative(Z):
    return (Z > 0).astype(float)
```

```python
def forward(X):
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2  # Linear output
    y_hat = Z2

    return Z1, A1, y_hat
```

```python
def mse(y, y_hat):
    return np.mean((y - y_hat) ** 2)
```

```python
def backward(X, y, Z1, A1, y_hat):
    m = X.shape[0]

    dZ2 = (2 / m) * (y_hat - y)
    dW2 = A1.T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)

    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * relu_derivative(Z1)

    dW1 = X.T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    return dW1, db1, dW2, db2

```

```python
lr = 0.01
epochs = 3000

for epoch in range(epochs):
    Z1, A1, y_hat = forward(X)
    loss = mse(y, y_hat)

    dW1, db1, dW2, db2 = backward(X, y, Z1, A1, y_hat)

    W1 -= lr * dW1
    b1 -= lr * db1
    W2 -= lr * dW2
    b2 -= lr * db2

    if epoch % 300 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.6f}")
```

```python
# Predict
_, _, y_pred_scaled = forward(X)

# Convert back to actual price
y_pred = y_pred_scaled * y_std + y_mean

print("Predicted prices:")
print(y_pred)
```
